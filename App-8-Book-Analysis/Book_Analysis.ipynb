# Load Book

```python
with open("miracle-in-the-andes.txt", "r", encoding="utf-8") as file:
    book = file.read()
```

```python
type(book)
```

# How many Chapters


### with string methods

```python
book.count("Chapter")
```

## With regrex

```python
import re
```

```python
pattern = re.compile("Chapter [0-9]+")
re.findall(pattern, book)
```

```python
pattern = re.compile("[^.]*[^a-zA-Z]love[^a-zA-Z][^.]*.")
findings = re.findall(pattern, book)
len(findings)
```
